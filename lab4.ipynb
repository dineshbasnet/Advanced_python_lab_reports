## **Theory**

### **1. SQLite**
- Lightweight, file-based relational database.
- Supports standard SQL queries.
- Does not require a server process; ideal for local development.
- **Features:** ACID compliance, cross-platform, serverless.

### **2. SQLAlchemy**
- Python ORM (Object Relational Mapper) for database abstraction.
- Maps Python classes to database tables.
- **Components:** Engine (connects database), Session (handles queries), Declarative Base (class-to-table mapping).

### **3. NoSQL (MongoDB)**
- Non-relational, document-oriented database.
- Stores data as JSON-like **BSON documents**.
- Flexible schema; ideal for hierarchical/unstructured data.
- **Features:** Schema-less, scalable, high-performance.

## **Implementation**

### **1. SQLite Example**

```python
import sqlite3

conn = sqlite3.connect('example_database.db')
cursor = conn.cursor()

cursor.execute('''
               CREATE TABLE IF NOT EXISTS students(
                   id INTEGER PRIMARY KEY,
                   sname TEXT,
                   age INTEGER,
                   grade REAL
               )
               ''')

cursor.execute("Insert into students(sname,age,grade) values ('Jonh',19,85.5)")

conn.commit()

cursor.execute("Select * from students")
print("Sqlite output:",cursor.fetchall())

conn.close()
```

### **2.SQLAlchemy ORM**

```python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base, sessionmaker

# Setup
Base = declarative_base()
engine = create_engine('sqlite:///lab_db.sqlite')
Session = sessionmaker(bind=engine)
session = Session()

# Define table as class
class Student(Base):
    __tablename__ = 'students'
    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)
    grade = Column(String)

# Create tables
Base.metadata.create_all(engine)

# Add a new student
new_student = Student(name='Harry', age=22, grade='A')
session.add(new_student)
session.commit()

# Query
for student in session.query(Student).all():
    print(student.name, student.age, student.grade)
```
