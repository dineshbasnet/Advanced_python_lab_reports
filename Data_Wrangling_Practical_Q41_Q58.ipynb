## Q48. Normalize and Standardize Numerical Features

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age':    [25, 35, 45, 20, 55],
    'Salary': [30000, 60000, 90000, 25000, 120000],
    'Score':  [70, 80, 65, 90, 75]
})
print("Original:")
print(df.to_string(index=False))

# Min-Max Normalization
def normalize(series):
    return (series - series.min()) / (series.max() - series.min())

# Z-score Standardization
def standardize(series):
    return (series - series.mean()) / series.std()

df_norm = df.apply(normalize)
df_std  = df.apply(standardize)

print("\nNormalized (Min-Max, range [0,1]):")
print(df_norm.round(4).to_string(index=False))

print("\nStandardized (Z-score, mean=0 std=1):")
print(df_std.round(4).to_string(index=False))

print("\nStandardized stats (mean should be ~0, std ~1):")
print(df_std.agg(['mean','std']).round(4))
```

## Q49. Reduce Memory Usage of a DataFrame

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'ID':       range(10000),
    'Age':      np.random.randint(18, 80, 10000),
    'Score':    np.random.uniform(0, 100, 10000),
    'Category': np.random.choice(['A','B','C','D'], 10000),
    'Flag':     np.random.randint(0, 2, 10000)
})

def mem_mb(d): return d.memory_usage(deep=True).sum() / 1024**2

print(f"Memory BEFORE optimization: {mem_mb(df):.4f} MB")
print(df.dtypes)

df_opt = df.copy()
df_opt['ID']       = df_opt['ID'].astype('int32')
df_opt['Age']      = df_opt['Age'].astype('int8')
df_opt['Score']    = df_opt['Score'].astype('float32')
df_opt['Category'] = df_opt['Category'].astype('category')
df_opt['Flag']     = df_opt['Flag'].astype('bool')

print(f"\nMemory AFTER optimization:  {mem_mb(df_opt):.4f} MB")
print(df_opt.dtypes)
reduction = (1 - mem_mb(df_opt)/mem_mb(df))*100
print(f"\nMemory reduced by: {reduction:.1f}%")
```

## Q50. Process a Large CSV File using Chunking

```python
import pandas as pd
import io

# Simulate a large CSV in memory
import random
rows = ["Category,Sales,Quantity"]
categories = ['Electronics','Clothing','Food','Sports','Books']
for _ in range(1000):
    rows.append(f"{random.choice(categories)},{random.randint(100,5000)},{random.randint(1,50)}")
csv_data = "\n".join(rows)

chunk_size = 200
total_sales = 0
total_rows  = 0
category_sales = {}

for i, chunk in enumerate(pd.read_csv(io.StringIO(csv_data), chunksize=chunk_size)):
    total_rows  += len(chunk)
    total_sales += chunk['Sales'].sum()
    for cat, grp in chunk.groupby('Category'):
        category_sales[cat] = category_sales.get(cat, 0) + grp['Sales'].sum()
    print(f"Chunk {i+1}: {len(chunk)} rows processed | Running total: {total_rows} rows")

print(f"\nTotal rows processed : {total_rows}")
print(f"Total sales (all chunks): {total_sales:,}")
print("\nSales by Category:")
for k, v in sorted(category_sales.items()):
    print(f"  {k:<15}: {v:>8,}")
```

## Q51. Reusable Data-Cleaning Function

```python
import pandas as pd
import numpy as np

def clean_dataframe(df, drop_dup=True, fill_numeric='mean', fill_cat='Unknown',
                    drop_thresh=None, convert_dates=None):
    df = df.copy()
    print(f"Initial shape: {df.shape}")

    # Remove duplicates
    if drop_dup:
        before = len(df)
        df.drop_duplicates(inplace=True)
        print(f"Duplicates removed: {before - len(df)}")

    # Drop cols with too many missing values
    if drop_thresh:
        df.dropna(thresh=int(len(df)*drop_thresh), axis=1, inplace=True)

    # Fill numeric missing values
    num_cols = df.select_dtypes(include=[np.number]).columns
    for col in num_cols:
        if df[col].isnull().any():
            val = df[col].mean() if fill_numeric == 'mean' else df[col].median()
            df[col].fillna(val, inplace=True)

    # Fill categorical missing values
    cat_cols = df.select_dtypes(include='object').columns
    for col in cat_cols:
        df[col].fillna(fill_cat, inplace=True)

    # Convert date columns
    if convert_dates:
        for col in convert_dates:
            df[col] = pd.to_datetime(df[col], errors='coerce')

    print(f"Final shape: {df.shape}")
    print(f"Missing values remaining: {df.isnull().sum().sum()}")
    return df

# Test the function
raw = pd.DataFrame({
    'Name':   ['Alice', 'Bob', None, 'Alice', 'Eve'],
    'Age':    [25, np.nan, 30, 25, 28],
    'Salary': [50000, 60000, np.nan, 50000, 45000],
    'Joined': ['2020-01-10', '2019-06-15', None, '2020-01-10', '2021-03-22']
})
print("Raw data:")
print(raw)
print()
cleaned = clean_dataframe(raw, convert_dates=['Joined'])
print("\nCleaned data:")
print(cleaned)
```

## Q52. Simple Data-Cleaning Pipeline using Pandas

```python
import pandas as pd
import numpy as np

# Raw messy dataset
raw_data = pd.DataFrame({
    'customer_id': [101, 102, 103, 103, 104, 105],
    'age':         [25, np.nan, 200, 22, 35, np.nan],
    'purchase':    [500.0, 1200.0, np.nan, 800.0, np.nan, 950.0],
    'category':    ['Electronics', 'clothing', None, 'Electronics', 'FOOD', 'Clothing'],
    'date':        ['2023-01-15', '2023-02-10', '2023-03-05', '2023-03-05', '2023-04-20', '2023-05-18']
})

def pipeline(df):
    df = df.copy()

    # Step 1: Remove duplicates
    df.drop_duplicates(inplace=True)
    print("Step 1 - After removing duplicates:", df.shape)

    # Step 2: Fix invalid age values
    df['age'] = df['age'].apply(lambda x: np.nan if (x and x > 120) else x)
    df['age'].fillna(df['age'].median(), inplace=True)

    # Step 3: Fill missing purchase with mean
    df['purchase'].fillna(df['purchase'].mean(), inplace=True)

    # Step 4: Standardize category (title case)
    df['category'].fillna('Unknown', inplace=True)
    df['category'] = df['category'].str.title()

    # Step 5: Parse dates
    df['date'] = pd.to_datetime(df['date'])

    # Step 6: Reset index
    df.reset_index(drop=True, inplace=True)

    print("Step 2-6 - After cleaning:")
    return df

print("BEFORE:")
print(raw_data)
print()
clean = pipeline(raw_data)
print()
print("AFTER:")
print(clean)
```

## Q55. RFM (Recency, Frequency, Monetary) Analysis

```python
import pandas as pd
import numpy as np
from datetime import datetime

np.random.seed(42)
n = 200
customers_ids = [f'C{str(i).zfill(3)}' for i in range(1, 31)]
df = pd.DataFrame({
    'customer_id': np.random.choice(customers_ids, n),
    'order_date':  pd.date_range('2022-01-01', periods=n, freq='2D').tolist(),
    'amount':      np.random.uniform(50, 500, n).round(2)
})

snapshot_date = df['order_date'].max() + pd.Timedelta(days=1)

rfm = df.groupby('customer_id').agg(
    Recency   = ('order_date', lambda x: (snapshot_date - x.max()).days),
    Frequency = ('order_date', 'count'),
    Monetary  = ('amount',     'sum')
).reset_index()

# Score each metric 1-4
for col, ascending in [('Recency', False), ('Frequency', True), ('Monetary', True)]:
    rfm[f'{col}_Score'] = pd.qcut(rfm[col], 4,
                                   labels=[1,2,3,4] if ascending else [4,3,2,1],
                                   duplicates='drop')

rfm['RFM_Score'] = (rfm['Recency_Score'].astype(int) +
                    rfm['Frequency_Score'].astype(int) +
                    rfm['Monetary_Score'].astype(int))

def segment(score):
    if score >= 10: return 'Champions'
    elif score >= 7: return 'Loyal'
    elif score >= 5: return 'At Risk'
    else: return 'Lost'

rfm['Segment'] = rfm['RFM_Score'].apply(segment)

print("RFM Analysis (top 10 customers):")
print(rfm.sort_values('RFM_Score', ascending=False).head(10).to_string(index=False))
print("\nSegment Distribution:")
print(rfm['Segment'].value_counts().to_string())
```

## Q56. Compute Customer Lifetime Value (CLV)

```python
import pandas as pd
import numpy as np

np.random.seed(0)
df = pd.DataFrame({
    'customer_id':    np.repeat([f'C{i:03d}' for i in range(1, 21)], 5),
    'order_date':     pd.date_range('2021-01-01', periods=100, freq='30D'),
    'purchase_amount':np.random.uniform(100, 1000, 100).round(2)
})

# CLV = Average Purchase Value × Purchase Frequency × Customer Lifespan
clv = df.groupby('customer_id').agg(
    Total_Revenue  = ('purchase_amount', 'sum'),
    Avg_Purchase   = ('purchase_amount', 'mean'),
    Num_Purchases  = ('purchase_amount', 'count'),
    First_Purchase = ('order_date',      'min'),
    Last_Purchase  = ('order_date',      'max')
).reset_index()

clv['Lifespan_Days'] = (clv['Last_Purchase'] - clv['First_Purchase']).dt.days + 1
clv['Purchase_Rate'] = clv['Num_Purchases'] / (clv['Lifespan_Days'] / 30)  # per month
clv['CLV'] = (clv['Avg_Purchase'] * clv['Purchase_Rate'] * 12).round(2)    # annual CLV

print("Customer Lifetime Value (CLV) — Top 10:")
print(clv[['customer_id','Total_Revenue','Avg_Purchase','Num_Purchases','CLV']]
      .sort_values('CLV', ascending=False).head(10).to_string(index=False))

print(f"\nAverage CLV across all customers: ${clv['CLV'].mean():,.2f}")
print(f"Highest CLV: ${clv['CLV'].max():,.2f}")
```

## Q58. End-to-End Data Pipeline using Python

```python
import pandas as pd
import numpy as np
import sqlite3, json, io, os

print("=" * 60)
print("   END-TO-END DATA PIPELINE")
print("=" * 60)

# ── STAGE 1: INGEST ──────────────────────────────────────────
print("\n[STAGE 1] DATA INGESTION")
csv_data = "cust_id,name,age,city\n1,Alice,25,NYC\n2,Bob,30,LA\n3,Carol,,Chicago\n4,David,28,NYC\n2,Bob,30,LA"
df_customers = pd.read_csv(io.StringIO(csv_data))

api_json = json.dumps([
    {"cust_id":1,"product":"Laptop","amount":1200,"date":"2023-01-15"},
    {"cust_id":2,"product":"Phone", "amount":800, "date":"2023-02-10"},
    {"cust_id":3,"product":"Tablet","amount":600, "date":"2023-03-05"},
    {"cust_id":4,"product":"Watch", "amount":350, "date":"2023-04-20"},
    {"cust_id":1,"product":"Mouse", "amount":50,  "date":"2023-05-01"},
])
df_orders = pd.DataFrame(json.loads(api_json))
print(f"  Customers ingested: {len(df_customers)} rows")
print(f"  Orders ingested   : {len(df_orders)} rows")

# ── STAGE 2: CLEAN ───────────────────────────────────────────
print("\n[STAGE 2] DATA CLEANING")
df_customers.drop_duplicates(inplace=True)
df_customers['age'].fillna(df_customers['age'].median(), inplace=True)
df_customers['age'] = df_customers['age'].astype(int)
df_orders['date'] = pd.to_datetime(df_orders['date'])
print(f"  After dedup customers: {len(df_customers)} rows")
print(f"  Missing values: {df_customers.isnull().sum().sum()}")

# ── STAGE 3: TRANSFORM ───────────────────────────────────────
print("\n[STAGE 3] DATA TRANSFORMATION")
df_merged = pd.merge(df_customers, df_orders, on='cust_id', how='left')
df_merged['amount'].fillna(0, inplace=True)

# Aggregate per customer
summary = df_merged.groupby(['cust_id','name','city']).agg(
    total_spent  = ('amount', 'sum'),
    num_orders   = ('amount', 'count'),
    avg_order    = ('amount', 'mean'),
    last_purchase= ('date',   'max')
).reset_index()

summary['customer_value'] = pd.cut(
    summary['total_spent'],
    bins=[-1, 500, 1000, float('inf')],
    labels=['Low','Medium','High']
)
print("  Transformed summary:")
print(summary.to_string(index=False))

# ── STAGE 4: STORE ───────────────────────────────────────────
print("\n[STAGE 4] DATA STORAGE")
conn = sqlite3.connect(':memory:')
summary.to_sql('customer_summary', conn, if_exists='replace', index=False)
print("  Stored in SQLite: customer_summary table")

csv_out = '/tmp/pipeline_output.csv'
summary.to_csv(csv_out, index=False)
print(f"  Exported CSV: {csv_out} ({os.path.getsize(csv_out)} bytes)")

# ── STAGE 5: VALIDATE ────────────────────────────────────────
print("\n[STAGE 5] VALIDATION")
result = pd.read_sql("SELECT customer_value, COUNT(*) as count, ROUND(SUM(total_spent),2) as revenue FROM customer_summary GROUP BY customer_value", conn)
print("  Revenue by Customer Value Segment:")
print(result.to_string(index=False))

conn.close()
print("\n Pipeline completed successfully!")
```
